```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
x=np.arange(1,11)
x
```

```python
y=2*x
y
```

```python
plt.plot(x,y)
plt.show()
```

```python
#Adding a title
plt.plot(x,y)
plt.title("first line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
# Adding coolour width and style of line
plt.plot(x,y,color="pink",linestyle=":",linewidth=5)
plt.title("first line plot")
plt.xlabel("x-axis")
plt.ylabel("y-axis")
plt.show()
```

```python
y1=2*x
y2=3*x
```

```python
plt.plot(x,y1,color="pink",linewidth=4)
plt.plot(x,y2,color="g",linewidth=5)
plt.grid()
plt.show()
```

```python
plt.subplot(1,2,1)
plt.plot(x,y1,color="black",linewidth=5)

plt.subplot(1,2,2)
plt.plot(x,y2,color="green",linewidth=5)

plt.grid(True)
plt.show()
```

```python
plt.subplot(2,1,1)
plt.plot(x,y1,color="black",linewidth=5)
plt.grid(True)

plt.subplot(2,1,2)
plt.plot(x,y2,color="green",linewidth=5)

plt.grid(True)
plt.show()
```
